```python
import os
import glob
import re
import pandas as pd
```

```python
# Define the folder path containing your extracted CSV files
folder_path = r"C:\Users\Juvair\Desktop\Green School\Green Educational Institute\Extracted_GEE\All_Data"

# Output file paths
output_csv = os.path.join(folder_path, "PrimarySchools_2026_All_Merged.csv")
output_excel = os.path.join(folder_path, "PrimarySchools_2026_All_Merged.xlsx")
```

```python
def extract_part_number(filename):
    """Extracts the integer part number from filenames like '...Part_ 1.csv' or '...Part_1.csv'."""
    match = re.search(r"Part_\s*(\d+)", filename, re.IGNORECASE)
    return int(match.group(1)) if match else 0


# Find all CSV files in the specified directory
all_files = glob.glob(os.path.join(folder_path, "*.csv"))

# Filter specifically for the Part_ 1 to 29 files
part_files = [f for f in all_files if "Part_" in os.path.basename(f)]

# Sort files numerically by Part number (1 to 29) rather than alphabetically
sorted_files = sorted(part_files, key=extract_part_number)

print(f"Found {len(sorted_files)} files to merge.")
print("Merge Order:")
for f in sorted_files:
    print(f"  - {os.path.basename(f)}")

# Read and combine CSVs sequentially
df_list = []
for file_path in sorted_files:
    # Read CSV (handling potential Unicode characters in school names or locations)
    df = pd.read_csv(file_path, encoding="utf-8-sig")
    df_list.append(df)

# Concatenate all DataFrames into one single DataFrame
combined_df = pd.concat(df_list, ignore_index=True)
```

```python

# Export to single CSV
print("\nExporting combined dataset to CSV...")
combined_df.to_csv(output_csv, index=False, encoding="utf-8-sig")
print(f"CSV saved successfully at:\n  {output_csv}")

# Export to single Excel (.xlsx) file
print("\nExporting combined dataset to Excel...")
combined_df.to_excel(output_excel, index=False, engine="openpyxl")
print(f"Excel saved successfully at:\n  {output_excel}")

print(f"\nCompleted! Total primary school records merged: {len(combined_df):,}")
```
